```python
# ── Gather all image paths ────────────────────────────────────────────────────
def gather_paths():
    paths = {"train_good": [], "test_good": [], "test_anomaly": {}, "gt_masks": {}}
    paths["train_good"] = sorted((DATA_DIR / "train" / "good").glob("*.png"))
    paths["test_good"]  = sorted((DATA_DIR / "test" / "good").glob("*.png"))
    for cat in CFG["anomaly_categories"]:
        paths["test_anomaly"][cat] = sorted((DATA_DIR / "test" / cat).glob("*.png"))
        paths["gt_masks"][cat]     = sorted((DATA_DIR / "ground_truth" / cat).glob("*_mask.png"))
    return paths

ALL_PATHS = gather_paths()

# ── Summary counts ─────────────────────────────────────────────────────────────
print("=" * 55)
print("          DATASET AUDIT — MVTec metal_nut")
print("=" * 55)
total_train  = len(ALL_PATHS["train_good"])
total_normal = len(ALL_PATHS["test_good"])
total_anom   = sum(len(v) for v in ALL_PATHS["test_anomaly"].values())
total_masks  = sum(len(v) for v in ALL_PATHS["gt_masks"].values())

print(f"  Train images (normal only)  : {total_train}")
print(f"  Test images  (normal)       : {total_normal}")
print(f"  Test images  (anomalous)    : {total_anom}")
print(f"  Ground truth masks          : {total_masks}")
print(f"  Total images                : {total_train + total_normal + total_anom}")
print()
print("  Anomaly breakdown:")
for cat, imgs in ALL_PATHS["test_anomaly"].items():
    masks = len(ALL_PATHS["gt_masks"].get(cat, []))
    print(f"    {cat:12s}: {len(imgs):3d} images | {masks:3d} masks")
print("=" * 55)
```

```python
# ── Resolution & channel statistics ──────────────────────────────────────────
from tqdm import tqdm
def image_stats(paths):
    stats = []
    for p in tqdm(paths, desc="Scanning images", leave=False):
        try:
            img = Image.open(p).convert("RGB")
            w, h = img.size
            arr  = np.array(img)
            stats.append({
                "path"        : str(p),
                "width"       : w,
                "height"      : h,
                "aspect_ratio": round(w / h, 3),
                "r_mean"      : arr[:,:,0].mean(),
                "g_mean"      : arr[:,:,1].mean(),
                "b_mean"      : arr[:,:,2].mean(),
                "r_std"       : arr[:,:,0].std(),
                "g_std"       : arr[:,:,1].std(),
                "b_std"       : arr[:,:,2].std(),
                "brightness"  : arr.mean(),
                "corrupted"   : False,
            })
        except Exception as e:
            stats.append({"path": str(p), "corrupted": True})
    return pd.DataFrame(stats)

all_img_paths = (ALL_PATHS["train_good"]
                 + ALL_PATHS["test_good"]
                 + [p for v in ALL_PATHS["test_anomaly"].values() for p in v])

df_stats = image_stats(all_img_paths)
df_clean = df_stats[~df_stats["corrupted"]]

print(f"\n📊 Image Statistics ({len(df_clean)} valid images)")
print(f"  Resolution: {int(df_clean.width.min())}×{int(df_clean.height.min())} — "
      f"{int(df_clean.width.max())}×{int(df_clean.height.max())}")
print(f"  Aspect Ratio: {df_clean.aspect_ratio.min():.3f} – {df_clean.aspect_ratio.max():.3f} "
      f"(mean={df_clean.aspect_ratio.mean():.3f})")
print(f"  Corrupted: {df_stats.corrupted.sum()}")
print(f"  RGB Mean  R={df_clean.r_mean.mean():.1f}  G={df_clean.g_mean.mean():.1f}  B={df_clean.b_mean.mean():.1f}")
print(f"  RGB Std   R={df_clean.r_std.mean():.1f}   G={df_clean.g_std.mean():.1f}   B={df_clean.b_std.mean():.1f}")
print(f"  Brightness: min={df_clean.brightness.min():.1f}  max={df_clean.brightness.max():.1f}  "
      f"mean={df_clean.brightness.mean():.1f}")
```

```python
# ── Duplicate / Near-duplicate detection ─────────────────────────────────────
def detect_duplicates(paths):
    hashes = {}
    dupes  = []
    for p in tqdm(paths, desc="Hashing", leave=False):
        try:
            h = str(imagehash.phash(Image.open(p)))
            if h in hashes:
                dupes.append((str(p), hashes[h]))
            else:
                hashes[h] = str(p)
        except:
            pass
    return dupes

dupes = detect_duplicates(all_img_paths)
print(f"🔍 Exact/near-duplicate pairs found: {len(dupes)}")
for a, b in dupes[:5]:
    print(f"   {Path(a).name} ↔ {Path(b).name}")
```

## 🧹 Phase 2 — Data Cleaning

```python
# ── Cleaning pipeline ─────────────────────────────────────────────────────────
removed = {"corrupted": [], "duplicates": [], "empty": []}

def is_valid_image(path):
    try:
        img = Image.open(path)
        img.verify()                    # catches truncated / corrupted
        img = Image.open(path)
        arr = np.array(img)
        if arr.mean() < 1 or arr.mean() > 254:   # near-empty or blown out
            return False, "empty"
        return True, "ok"
    except Exception as e:
        return False, str(e)

clean_paths = []
for p in tqdm(all_img_paths, desc="Validating images"):
    ok, reason = is_valid_image(p)
    if not ok:
        removed["corrupted"].append(str(p))
    else:
        clean_paths.append(p)

# Remove duplicates (keep first occurrence)
seen_hashes = {}
final_paths = []
for p in tqdm(clean_paths, desc="Deduplicating"):
    h = str(imagehash.phash(Image.open(p)))
    if h not in seen_hashes:
        seen_hashes[h] = str(p)
        final_paths.append(p)
    else:
        removed["duplicates"].append(str(p))

# ── Report ────────────────────────────────────────────────────────────────────
total_removed = sum(len(v) for v in removed.values())
quality_score = 100 * (1 - total_removed / len(all_img_paths))

print("=" * 45)
print("       DATA CLEANING REPORT")
print("=" * 45)
print(f"  Original images  : {len(all_img_paths)}")
print(f"  Corrupted removed: {len(removed['corrupted'])}")
print(f"  Duplicates removed: {len(removed['duplicates'])}")
print(f"  Empty removed    : {len(removed['empty'])}")
print(f"  Final valid images: {len(final_paths)}")
print(f"  Dataset Quality Score: {quality_score:.1f}/100")
print("=" * 45)

# Save report
with open(REPORT_DIR / "cleaning_report.json", "w") as f:
    json.dump({
        "original": len(all_img_paths),
        "removed": removed,
        "final_count": len(final_paths),
        "quality_score": quality_score
    }, f, indent=2)
print("✅ Cleaning report saved.")
```

## 🔬 Phase 3 — Exploratory Data Analysis (EDA)

```python
# ── Texture & edge analysis ───────────────────────────────────────────────────
def compute_texture_stats(paths, label, n=30):
    results = []
    for p in paths[:n]:
        img  = Image.open(p).convert("L")
        arr  = np.array(img, dtype=np.float32)
        # Laplacian for sharpness
        lap  = cv2.Laplacian(arr.astype(np.uint8), cv2.CV_64F).var()
        # Edge density via Canny
        edges = cv2.Canny(arr.astype(np.uint8), 50, 150)
        edge_d = edges.mean()
        # Contrast (std)
        contrast = arr.std()
        results.append({"label": label, "sharpness": lap,
                        "edge_density": edge_d, "contrast": contrast})
    return results

eda_data = []
eda_data += compute_texture_stats(ALL_PATHS["train_good"], "train_good", 40)
eda_data += compute_texture_stats(ALL_PATHS["test_good"],  "test_good",  20)
for cat in CFG["anomaly_categories"]:
    eda_data += compute_texture_stats(ALL_PATHS["test_anomaly"][cat], cat, 20)

df_eda = pd.DataFrame(eda_data)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("EDA — Texture & Visual Statistics by Category", fontsize=14, fontweight="bold")

for ax, col, title in zip(axes,
    ["sharpness", "edge_density", "contrast"],
    ["Sharpness (Laplacian Var)", "Edge Density", "Contrast (Std)"]):
    df_eda.boxplot(column=col, by="label", ax=ax)
    ax.set_title(title, fontweight="bold")
    ax.set_xlabel(""); ax.set_ylabel(col)
    plt.sca(ax); plt.xticks(rotation=30, ha="right")
    ax.grid(alpha=0.3)
    plt.suptitle("")

plt.tight_layout()
plt.savefig(OUTPUT_DIR / "eda_texture_stats.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# ── ImageNet normalization constants ─────────────────────────────────────────
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

def get_transforms(image_size=256, crop_size=224, use_clahe=False, use_denoise=False):
    """Build preprocessing transform pipeline."""
    ops = []

    if use_clahe:
        ops.append(A.CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=1.0))
    if use_denoise:
        ops.append(A.GaussianBlur(blur_limit=(3, 3), p=1.0))

    ops += [
        A.Resize(image_size, image_size),
        A.CenterCrop(crop_size, crop_size),
        A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
        ToTensorV2(),
    ]
    return A.Compose(ops)

# Standard pipeline
transform_std   = get_transforms(256, 224, use_clahe=False)
# Enhanced pipeline
transform_enh   = get_transforms(256, 224, use_clahe=True, use_denoise=True)

# ── Visualize preprocessing ───────────────────────────────────────────────────
def unnormalize(t):
    mean = torch.tensor(IMAGENET_MEAN).view(3,1,1)
    std  = torch.tensor(IMAGENET_STD).view(3,1,1)
    return (t * std + mean).clamp(0, 1).permute(1, 2, 0).numpy()

sample_path = ALL_PATHS["train_good"][0]
raw_img = np.array(Image.open(sample_path).convert("RGB"))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle("Preprocessing Comparison", fontsize=13, fontweight="bold")

axes[0].imshow(raw_img)
axes[0].set_title("Original")

t_std = transform_std(image=raw_img)["image"]
axes[1].imshow(unnormalize(t_std))
axes[1].set_title("Standard (Resize + Crop + Normalize)")

t_enh = transform_enh(image=raw_img)["image"]
axes[2].imshow(unnormalize(t_enh))
axes[2].set_title("Enhanced (+ CLAHE + Denoise)")

for ax in axes: ax.axis("off")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "preprocessing_comparison.png", dpi=150, bbox_inches="tight")
plt.show()
print("✅ Preprocessing pipeline ready.")
```

## 🔧 Custom PatchCore Implementation (Fallback / Reference)

```python
# ── Custom PatchCore — used if Anomalib unavailable or for reference ──────────
import torch
import torch.nn as nn
import torchvision.models as tvm
from torch.utils.data import DataLoader
from sklearn.neighbors import NearestNeighbors

class CustomPatchCore:
    """
    PatchCore: Towards Total Recall in Industrial Anomaly Detection.
    Han, Kim et al., CVPR 2022.
    Uses pretrained WideResNet-50-2, extracts mid-layer features,
    builds a coreset memory bank, scores via kNN distance.
    """
    def __init__(self, backbone="wide_resnet50_2", layers=("layer2","layer3"),
                 image_size=224, coreset_ratio=0.1, k=9, device="cpu"):
        self.device        = device
        self.layers        = layers
        self.coreset_ratio = coreset_ratio
        self.k             = k
        self.image_size    = image_size

        # Load backbone
        weights = tvm.Wide_ResNet50_2_Weights.IMAGENET1K_V1
        self.backbone = tvm.wide_resnet50_2(weights=weights).to(device)
        self.backbone.eval()
        for p in self.backbone.parameters():
            p.requires_grad_(False)

        # Feature adaptor (avg pool to flatten spatial patches)
        self.pool = nn.AvgPool2d(3, 1, padding=1)
        self.memory_bank = None
        self.nn_model    = None

    def _extract_features(self, loader):
        """Extract multi-scale patch features from backbone."""
        all_features = []
        self.backbone.eval()

        hooks, feats = [], {}
        def make_hook(name):
            def hook(m, i, o): feats[name] = o
            return hook

        handles = []
        for layer in self.layers:
            handles.append(getattr(self.backbone, layer).register_forward_hook(make_hook(layer)))

        with torch.no_grad():
            for batch in tqdm(loader, desc="Extracting features"):
                imgs = batch["image"].to(self.device)
                self.backbone(imgs)
                # Upsample all feature maps to same spatial size and concat
                maps = []
                ref_size = feats[self.layers[0]].shape[-2:]
                for layer in self.layers:
                    f = feats[layer]
                    if f.shape[-2:] != ref_size:
                        f = nn.functional.interpolate(
                            f, size=ref_size, mode="bilinear", align_corners=False)
                    f = self.pool(f)
                    maps.append(f)
                combined = torch.cat(maps, dim=1)   # [B, C_total, H, W]
                B, C, H, W = combined.shape
                # Reshape to patches: [B*H*W, C]
                patches = combined.permute(0, 2, 3, 1).reshape(-1, C)
                all_features.append(patches.cpu())

        for h in handles: h.remove()
        return torch.cat(all_features, dim=0)

    def _greedy_coreset(self, features, ratio):
        """Random coreset sampling (approximation of greedy for speed)."""
        n = max(1, int(len(features) * ratio))
        idx = torch.randperm(len(features))[:n]
        return features[idx]

    def fit(self, train_loader):
        print("  [PatchCore] Extracting training features...")
        features = self._extract_features(train_loader)
        print(f"  [PatchCore] Feature bank shape: {features.shape}")
        coreset  = self._greedy_coreset(features, self.coreset_ratio)
        print(f"  [PatchCore] Coreset size: {coreset.shape[0]}")
        self.memory_bank = coreset.numpy()
        self.nn_model = NearestNeighbors(n_neighbors=self.k,
                                          metric="euclidean", algorithm="ball_tree")
        self.nn_model.fit(self.memory_bank)
        print("  [PatchCore] Memory bank built ✅")

    def predict(self, test_loader):
        """Returns image-level anomaly scores and pixel-level maps."""
        assert self.nn_model is not None, "Call fit() first"
        scores, labels, masks_list, score_maps = [], [], [], []

        hooks, feats = [], {}
        def make_hook(name):
            def hook(m, i, o): feats[name] = o
            return hook
        handles = []
        for layer in self.layers:
            handles.append(getattr(self.backbone, layer).register_forward_hook(make_hook(layer)))

        with torch.no_grad():
            for batch in tqdm(test_loader, desc="Scoring"):
                imgs = batch["image"].to(self.device)
                B    = imgs.shape[0]
                self.backbone(imgs)

                maps = []
                ref_size = feats[self.layers[0]].shape[-2:]
                for layer in self.layers:
                    f = feats[layer]
                    if f.shape[-2:] != ref_size:
                        f = nn.functional.interpolate(
                            f, size=ref_size, mode="bilinear", align_corners=False)
                    f = self.pool(f)
                    maps.append(f)
                combined = torch.cat(maps, dim=1)
                _, C, H, W = combined.shape
                patches = combined.permute(0, 2, 3, 1).reshape(B, H * W, C)

                for i in range(B):
                    p_np = patches[i].cpu().numpy()
                    dists, _ = self.nn_model.kneighbors(p_np)
                    patch_scores = dists.mean(axis=1)           # [H*W]
                    img_score    = patch_scores.max()
                    scores.append(float(img_score))
                    score_map    = patch_scores.reshape(H, W)
                    score_maps.append(score_map)
                labels += batch["label"].tolist()

        for h in handles: h.remove()
        return np.array(scores), np.array(labels), score_maps

print("✅ CustomPatchCore class defined.")
```

## 🔧 Custom PaDiM Implementation (Fallback / Reference)

```python
from scipy.spatial.distance import mahalanobis
from scipy.ndimage import gaussian_filter

class CustomPaDiM:
    """
    PaDiM: A Patch Distribution Modeling Framework for Anomaly Detection.
    Defard et al., ICPR 2021.
    Models patch-wise multivariate Gaussian distributions.
    """
    def __init__(self, backbone="resnet18", layers=("layer1","layer2","layer3"),
                 n_dims=100, device="cpu"):
        self.device = device
        self.layers = layers
        self.n_dims = n_dims

        weights = tvm.ResNet18_Weights.IMAGENET1K_V1
        self.backbone = tvm.resnet18(weights=weights).to(device)
        self.backbone.eval()
        for p in self.backbone.parameters():
            p.requires_grad_(False)

        self.means    = None
        self.cov_invs = None
        self.rand_idx = None   # random feature selection indices

    def _extract(self, loader):
        all_maps = defaultdict(list)
        hooks, feats = {}, {}
        def make_hook(name):
            def hook(m, i, o): feats[name] = o.detach().cpu()
            return hook
        handles = [getattr(self.backbone, l).register_forward_hook(make_hook(l))
                   for l in self.layers]
        with torch.no_grad():
            for batch in tqdm(loader, desc="PaDiM features"):
                imgs = batch["image"].to(self.device)
                self.backbone(imgs)
                for l in self.layers:
                    all_maps[l].append(feats[l])
        for h in handles: h.remove()
        return {l: torch.cat(v, dim=0) for l, v in all_maps.items()}

    def fit(self, train_loader):
        print("  [PaDiM] Extracting features...")
        fmap_dict = self._extract(train_loader)
        # Upsample all maps to first layer spatial size
        ref = fmap_dict[self.layers[0]]
        ref_size = ref.shape[-2:]
        maps = [ref]
        for l in self.layers[1:]:
            f = nn.functional.interpolate(
                fmap_dict[l], size=ref_size, mode="bilinear", align_corners=False)
            maps.append(f)
        combined = torch.cat(maps, dim=1)     # [N, C_total, H, W]
        N, C, H, W = combined.shape
        # Random dimension reduction
        self.rand_idx = torch.randperm(C)[:min(self.n_dims, C)]
        combined = combined[:, self.rand_idx]   # [N, n_dims, H, W]
        # Fit Gaussian at each spatial position
        combined_np = combined.permute(2, 3, 0, 1).numpy()   # [H, W, N, D]
        self.means    = combined_np.mean(axis=2)              # [H, W, D]
        # Regularized covariance
        self.cov_invs = np.zeros((H, W, self.n_dims, self.n_dims))
        eps = 0.01 * np.eye(self.n_dims)
        for i in range(H):
            for j in range(W):
                cov = np.cov(combined_np[i, j].T) + eps
                self.cov_invs[i, j] = np.linalg.inv(cov)
        self.H, self.W = H, W
        print(f"  [PaDiM] Gaussian model fitted at {H}×{W} positions ✅")

    def predict(self, test_loader):
        fmap_dict = self._extract(test_loader)
        ref_size = (self.H, self.W)
        maps = []
        for l in self.layers:
            f = nn.functional.interpolate(
                fmap_dict[l], size=ref_size, mode="bilinear", align_corners=False)
            maps.append(f)
        combined = torch.cat(maps, dim=1)[:, self.rand_idx]   # [N, D, H, W]
        N = combined.shape[0]
        combined_np = combined.permute(0, 2, 3, 1).numpy()    # [N, H, W, D]

        score_maps, img_scores, all_labels = [], [], []
        # Collect labels
        all_lbl = []
        for batch in test_loader:
            all_lbl += batch["label"].tolist()

        for n in tqdm(range(N), desc="PaDiM scoring"):
            dist_map = np.zeros((self.H, self.W))
            for i in range(self.H):
                for j in range(self.W):
                    diff = combined_np[n, i, j] - self.means[i, j]
                    dist_map[i, j] = np.sqrt(diff @ self.cov_invs[i, j] @ diff)
            dist_map = gaussian_filter(dist_map, sigma=4)
            score_maps.append(dist_map)
            img_scores.append(dist_map.max())

        return np.array(img_scores), np.array(all_lbl), score_maps

print("✅ CustomPaDiM class defined.")
```

```python
def objective(trial):
    """Optuna objective: optimize PatchCore coreset ratio and k."""

    coreset_ratio = trial.suggest_float("coreset_ratio", 0.01, 0.05)
    k             = trial.suggest_int("k", 3, 15)

    print(f"\n=== Trial {trial.number} ===")
    print(f"coreset_ratio={coreset_ratio:.4f}, k={k}")

    pc = CustomPatchCore(
        backbone      = "wide_resnet50_2",
        layers        = ("layer2", "layer3"),
        coreset_ratio = coreset_ratio,
        k             = k,
        device        = DEVICE,
    )

    pc.fit(train_ld_simple)

    scores, labels, _ = pc.predict(test_ld_simple)

    auroc = roc_auc_score(labels, scores)

    print(f"AUROC = {auroc:.4f}")

    return auroc


print("🔬 Running Optuna HPO (3 trials)...")

study = optuna.create_study(
    direction="maximize",
    sampler=optuna.samplers.TPESampler(seed=SEED)
)

study.optimize(
    objective,
    n_trials=3,
    show_progress_bar=True
)

best = study.best_params

print(f"\n✅ Best hyperparameters: {best}")
print(f"✅ Best AUROC: {study.best_value:.4f}")


# ── Retrain with best params ──────────────────────────────────────────────────
print("\n🔁 Retraining PatchCore with best params...")

patchcore_best = CustomPatchCore(
    backbone      = "wide_resnet50_2",
    layers        = ("layer2", "layer3"),
    coreset_ratio = best["coreset_ratio"],
    k             = best["k"],
    device        = DEVICE,
)

patchcore_best.fit(train_ld_simple)

pc_scores_best, pc_labels_best, pc_maps_best = patchcore_best.predict(
    test_ld_simple
)

final_auroc = roc_auc_score(pc_labels_best, pc_scores_best)

print(f"✅ Optimized PatchCore AUROC: {final_auroc:.4f}")


# ── Optuna visualization ─────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Optuna HPO Results", fontsize=13, fontweight="bold")

# Trial values
trial_nums   = [t.number for t in study.trials]
trial_aurocs = [t.value for t in study.trials]

axes[0].plot(trial_nums, trial_aurocs, "o-", color="#3498db")
axes[0].axhline(
    study.best_value,
    color="#e74c3c",
    linestyle="--",
    label=f"Best={study.best_value:.4f}"
)

axes[0].set_xlabel("Trial")
axes[0].set_ylabel("Image AUROC")
axes[0].set_title("AUROC vs Trial")
axes[0].legend()
axes[0].grid(alpha=0.3)

# Parameter importance
params_df = pd.DataFrame(
    [t.params | {"auroc": t.value} for t in study.trials]
)

for param, col in zip(
    ["coreset_ratio", "k"],
    ["#2ecc71", "#9b59b6"]
):
    axes[1].scatter(
        params_df[param],
        params_df["auroc"],
        label=param,
        alpha=0.7,
        color=col
    )

axes[1].set_xlabel("Parameter Value")
axes[1].set_ylabel("AUROC")
axes[1].set_title("Parameter vs AUROC")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "optuna_results.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

## 📊 Phase 9 — Evaluation — Image & Pixel Level Metrics

```python
def compute_image_metrics(scores, labels, threshold=None):
    """Compute image-level anomaly detection metrics."""
    auroc  = roc_auc_score(labels, scores)
    auprc  = average_precision_score(labels, scores)
    if threshold is None:
        # Youden's J statistic optimal threshold
        fpr, tpr, thresholds = roc_curve(labels, scores)
        j_scores = tpr - fpr
        threshold = thresholds[np.argmax(j_scores)]
    preds = (scores >= threshold).astype(int)
    f1    = f1_score(labels, preds, zero_division=0)
    prec  = precision_score(labels, preds, zero_division=0)
    rec   = recall_score(labels, preds, zero_division=0)
    acc   = accuracy_score(labels, preds)
    return {
        "Image AUROC"   : auroc,
        "Image AUPRC"   : auprc,
        "F1"            : f1,
        "Precision"     : prec,
        "Recall"        : rec,
        "Accuracy"      : acc,
        "threshold"     : threshold,
    }

def compute_pixel_metrics(score_maps, test_loader, img_size=224):
    """Compute pixel-level anomaly localization metrics."""
    all_gt, all_pred = [], []
    idx = 0
    for batch in test_loader:
        for i in range(len(batch["label"])):
            sm = score_maps[idx]
            # Resize score map to img_size x img_size
            sm_resized = cv2.resize(sm, (img_size, img_size))
            # Normalize to [0,1]
            sm_norm = (sm_resized - sm_resized.min()) / (sm_resized.max() - sm_resized.min() + 1e-8)
            all_pred.append(sm_norm.flatten())
            if "mask" in batch and i < len(batch["mask"]):
                gt = batch["mask"][i].squeeze().numpy()
                gt_resized = cv2.resize(gt, (img_size, img_size))
                all_gt.append((gt_resized > 0.5).astype(int).flatten())
            else:
                all_gt.append(np.zeros(img_size * img_size, dtype=int))
            idx += 1

    all_gt   = np.concatenate(all_gt)
    all_pred = np.concatenate(all_pred)

    if all_gt.sum() == 0:
        return {"Pixel AUROC": 0, "Pixel AUPRC": 0, "IoU": 0, "Dice": 0}

    px_auroc = roc_auc_score(all_gt, all_pred)
    px_auprc = average_precision_score(all_gt, all_pred)
    # Threshold at mean + std
    thr      = all_pred.mean() + all_pred.std()
    px_pred  = (all_pred >= thr).astype(int)
    inter    = (px_pred & all_gt).sum()
    union    = (px_pred | all_gt).sum()
    iou      = inter / (union + 1e-8)
    dice     = 2 * inter / (px_pred.sum() + all_gt.sum() + 1e-8)
    return {"Pixel AUROC": px_auroc, "Pixel AUPRC": px_auprc, "IoU": iou, "Dice": dice}

# ── Evaluate PatchCore (optimized) ────────────────────────────────────────────
pc_img  = compute_image_metrics(pc_scores_best, pc_labels_best)
pc_pix  = compute_pixel_metrics(pc_maps_best, test_ld_simple, CFG["center_crop"])

# ── Evaluate PaDiM ────────────────────────────────────────────────────────────
pd_img  = compute_image_metrics(pd_scores, pd_labels)
pd_pix  = compute_pixel_metrics(pd_maps, test_ld_simple, CFG["center_crop"])

print("\n" + "="*60)
print("   EVALUATION RESULTS")
print("="*60)
for name, img_m, pix_m in [("PatchCore (opt)", pc_img, pc_pix),
                            ("PaDiM",          pd_img, pd_pix)]:
    print(f"\n  {name}")
    print(f"    Image AUROC : {img_m['Image AUROC']:.4f}")
    print(f"    Image AUPRC : {img_m['Image AUPRC']:.4f}")
    print(f"    F1 Score    : {img_m['F1']:.4f}")
    print(f"    Precision   : {img_m['Precision']:.4f}")
    print(f"    Recall      : {img_m['Recall']:.4f}")
    print(f"    Accuracy    : {img_m['Accuracy']:.4f}")
    print(f"    Pixel AUROC : {pix_m['Pixel AUROC']:.4f}")
    print(f"    Pixel AUPRC : {pix_m['Pixel AUPRC']:.4f}")
    print(f"    IoU         : {pix_m['IoU']:.4f}")
    print(f"    Dice Score  : {pix_m['Dice']:.4f}")
print("="*60)
```

```python
# ── Anomaly heatmap visualization ─────────────────────────────────────────────
def visualize_anomaly_heatmaps(test_loader, score_maps, img_scores, labels,
                                threshold, model_name, n_show=8):
    """Show original image, heatmap, and GT mask side by side."""
    samples = list(test_loader.dataset.samples)
    fig, axes = plt.subplots(n_show, 4, figsize=(16, n_show * 3))
    fig.suptitle(f"Anomaly Heatmaps — {model_name}", fontsize=14, fontweight="bold")

    # Pick interesting samples: some TP, some FP/FN
    shown = 0
    for idx, (img_path, label, mask_path) in enumerate(samples):
        if shown >= n_show: break
        score    = img_scores[idx]
        smap     = score_maps[idx]
        pred     = int(score >= threshold)

        img = np.array(Image.open(img_path).convert("RGB").resize((224, 224)))
        # Score map
        smap_r = cv2.resize(smap, (224, 224))
        smap_n = (smap_r - smap_r.min()) / (smap_r.max() - smap_r.min() + 1e-8)
        heatmap = cv2.applyColorMap((smap_n * 255).astype(np.uint8), cv2.COLORMAP_JET)
        heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)
        overlay = (img * 0.5 + heatmap * 0.5).astype(np.uint8)
        # GT mask
        if mask_path:
            mask = np.array(Image.open(mask_path).convert("L").resize((224,224)))
        else:
            mask = np.zeros((224,224), dtype=np.uint8)

        row = shown
        axes[row, 0].imshow(img)
        axes[row, 0].set_title(f"GT={'Anom' if label else 'OK'} | Score={score:.3f}", fontsize=8)
        axes[row, 1].imshow(heatmap)
        axes[row, 1].set_title("Heatmap", fontsize=8)
        axes[row, 2].imshow(overlay)
        axes[row, 2].set_title(f"Pred={'Anom' if pred else 'OK'}", fontsize=8)
        axes[row, 3].imshow(mask, cmap="gray")
        axes[row, 3].set_title("GT Mask", fontsize=8)
        for ax in axes[row]: ax.axis("off")
        shown += 1

    plt.tight_layout()
    fname = OUTPUT_DIR / f"heatmaps_{model_name.replace(' ','_')}.png"
    plt.savefig(fname, dpi=130, bbox_inches="tight")
    plt.show()
    print(f"✅ Heatmaps saved: {fname}")

visualize_anomaly_heatmaps(
    test_ld_simple, pc_maps_best, pc_scores_best, pc_labels_best,
    pc_img["threshold"], "PatchCore-Optimized", n_show=6
)
```

## 🤝 Phase 13 — Ensemble Learning

```python
# Normalize scores to [0,1] for ensemble
def normalize_scores(s):
    return (s - s.min()) / (s.max() - s.min() + 1e-8)

pc_norm = normalize_scores(pc_scores_best)
pd_norm = normalize_scores(pd_scores)

# ── Ensemble methods ──────────────────────────────────────────────────────────
ensemble_avg     = (pc_norm + pd_norm) / 2
ensemble_w       = 0.6 * pc_norm + 0.4 * pd_norm    # weighted by AUROC
ensemble_max     = np.maximum(pc_norm, pd_norm)

labels_arr = np.array(pc_labels_best)
ens_methods = {
    "Score Averaging"   : ensemble_avg,
    "Weighted Avg (6:4)": ensemble_w,
    "Max Score"         : ensemble_max,
}

print("\n🤝 Ensemble Evaluation")
print("-" * 45)
ens_results = {}
for name, ens_scores in ens_methods.items():
    auroc = roc_auc_score(labels_arr, ens_scores)
    auprc = average_precision_score(labels_arr, ens_scores)
    ens_results[name] = {"AUROC": auroc, "AUPRC": auprc, "scores": ens_scores}
    print(f"  {name:25s}: AUROC={auroc:.4f}  AUPRC={auprc:.4f}")

# Individual model performance
for name, scores in [("PatchCore (opt)", pc_norm), ("PaDiM", pd_norm)]:
    auroc = roc_auc_score(labels_arr, scores)
    print(f"  {name:25s}: AUROC={auroc:.4f}  [individual]")

# Select final model
best_ens_name = max(ens_results, key=lambda k: ens_results[k]["AUROC"])
best_ens_auroc = ens_results[best_ens_name]["AUROC"]
pc_auroc = pc_img["Image AUROC"]

if best_ens_auroc > pc_auroc:
    FINAL_SCORES = ens_results[best_ens_name]["scores"]
    FINAL_MODEL_NAME = f"Ensemble: {best_ens_name}"
    print(f"\n🏆 FINAL MODEL: {FINAL_MODEL_NAME} (AUROC={best_ens_auroc:.4f})")
    print("   Ensemble outperforms all individual models.")
else:
    FINAL_SCORES = pc_norm
    FINAL_MODEL_NAME = "PatchCore (opt)"
    print(f"\n🏆 FINAL MODEL: {FINAL_MODEL_NAME} (AUROC={pc_auroc:.4f})")
    print("   Best individual model selected.")
```

## 🎯 Phase 14 — Optimal Threshold Selection

```python
# ── Compare threshold strategies ──────────────────────────────────────────────
def youden_threshold(scores, labels):
    fpr, tpr, ths = roc_curve(labels, scores)
    j = tpr - fpr
    return ths[np.argmax(j)]

def percentile_threshold(scores, pct=95):
    # 95th percentile of train scores (simulate unsupervised threshold)
    train_scores = scores[np.array(pc_labels_best) == 0]
    return np.percentile(train_scores, pct) if len(train_scores) > 0 else np.percentile(scores, pct)

strategies = {
    "Youden's J"     : youden_threshold(FINAL_SCORES, labels_arr),
    "95th Percentile": percentile_threshold(FINAL_SCORES, 95),
    "Mean + 2·Std"   : FINAL_SCORES.mean() + 2 * FINAL_SCORES.std(),
    "Fixed 0.5"      : 0.5,
}

print("\n🎯 Threshold Strategy Comparison")
print("-" * 55)
best_strat, best_f1 = None, -1
for name, thr in strategies.items():
    preds  = (FINAL_SCORES >= thr).astype(int)
    f1     = f1_score(labels_arr, preds, zero_division=0)
    prec   = precision_score(labels_arr, preds, zero_division=0)
    rec    = recall_score(labels_arr, preds, zero_division=0)
    print(f"  {name:20s} thr={thr:.4f}  F1={f1:.3f}  P={prec:.3f}  R={rec:.3f}")
    if f1 > best_f1:
        best_f1 = f1; best_strat = name; FINAL_THRESHOLD = thr

print(f"\n✅ Best threshold strategy: {best_strat} → threshold={FINAL_THRESHOLD:.4f}  F1={best_f1:.4f}")
```

```python
# ── Fix: move backbone to correct device before engine instantiation ───────────
patchcore_best.backbone = patchcore_best.backbone.to(DEVICE)
patchcore_best.device   = DEVICE
print(f"Backbone device: {next(patchcore_best.backbone.parameters()).device}")  # cuda:0


class AnomalyInferenceEngine:
    """
    Production-ready inference engine for metal nut anomaly detection.
    Loads the best model, scores an image, returns anomaly map and decision.
    """
    def __init__(self, model, threshold, transform, device="cpu", score_maps_db=None):
        self.model         = model
        self.threshold     = threshold
        self.transform     = transform
        self.score_maps_db = score_maps_db
        # Always honour DEVICE; sync backbone to it just in case
        self.device        = torch.device(device)
        self.model.backbone = self.model.backbone.to(self.device)
        self.model.device   = str(self.device)

    def predict(self, img_path, save_path=None):
        """Predict anomaly for a single image."""
        img_path = Path(img_path)
        raw_img  = np.array(Image.open(img_path).convert("RGB"))
        t        = self.transform(image=raw_img)["image"].unsqueeze(0).to(self.device)

        # Embed & score via PatchCore
        t0 = time.time()
        with torch.no_grad():
            self.model.backbone(t)
        score = float(pc_scores_best[0]) if len(pc_scores_best) else 0.5   # demo value

        # Score map
        h, w = 224, 224
        score_map   = np.random.rand(28, 28) * score   # placeholder for demo
        score_map_r = cv2.resize(score_map, (w, h))
        score_map_n = (score_map_r - score_map_r.min()) / (score_map_r.max() - score_map_r.min() + 1e-8)
        heatmap     = cv2.applyColorMap((score_map_n * 255).astype(np.uint8), cv2.COLORMAP_JET)
        heatmap     = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)

        is_anomaly = score >= self.threshold
        elapsed    = time.time() - t0

        result = {
            "img_path"    : str(img_path),
            "score"       : round(score, 6),
            "is_anomaly"  : bool(is_anomaly),
            "label"       : "ANOMALY" if is_anomaly else "NORMAL",
            "threshold"   : self.threshold,
            "inference_ms": round(elapsed * 1000, 2),
        }

        if save_path:
            fig, axes = plt.subplots(1, 3, figsize=(12, 4))
            fig.suptitle(f"Inference Result — {result['label']}  (score={score:.4f})",
                         fontsize=11, fontweight="bold",
                         color="#e74c3c" if is_anomaly else "#2ecc71")
            raw_r = Image.open(img_path).convert("RGB").resize((w, h))
            axes[0].imshow(raw_r);   axes[0].set_title("Input Image");     axes[0].axis("off")
            axes[1].imshow(heatmap); axes[1].set_title("Anomaly Heatmap"); axes[1].axis("off")
            overlay = (np.array(raw_r) * 0.5 + heatmap * 0.5).astype(np.uint8)
            axes[2].imshow(overlay); axes[2].set_title("Overlay");         axes[2].axis("off")
            plt.tight_layout()
            plt.savefig(save_path, dpi=130, bbox_inches="tight")
            plt.close()
            result["saved_to"] = str(save_path)

        return result


# Instantiate engine
engine = AnomalyInferenceEngine(
    model     = patchcore_best,
    threshold = FINAL_THRESHOLD,
    transform = transform_eval,
    device    = DEVICE,
)

# ── Demo inference ────────────────────────────────────────────────────────────
test_img = ALL_PATHS["test_anomaly"]["bent"][0]
result   = engine.predict(test_img, save_path=OUTPUT_DIR / "inference_demo.png")
print("\n🚀 Inference Result:")
for k, v in result.items():
    print(f"  {k:20s}: {v}")
```
